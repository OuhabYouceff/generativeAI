```python
import torch
from transformers import CLIPTextModel, CLIPTokenizer
from diffusers import StableDiffusionPipeline, DiffusionPipeline, DPMSolverMultistepScheduler
from flask import Flask, request, send_file, jsonify
import io
from PIL import Image
import threading
from pyngrok import ngrok
from flask_cors import CORS
import cv2
import numpy as np

# Initialize Flask app
app = Flask(__name__)
CORS(app, resources={r"/*": {"origins": "*"}})

# Load the Stable Diffusion pipeline for image generation
image_pipe = StableDiffusionPipeline.from_pretrained("CompVis/stable-diffusion-v1-4", torch_dtype=torch.float16).to("cuda")

# Load the Diffusion pipeline for video generation
video_pipe = DiffusionPipeline.from_pretrained("damo-vilab/text-to-video-ms-1.7b", torch_dtype=torch.float16, variant="fp16")
video_pipe.scheduler = DPMSolverMultistepScheduler.from_config(video_pipe.scheduler.config)
video_pipe.enable_model_cpu_offload()

# Load the new Stable Diffusion 2.1 base model
model_id = "stabilityai/stable-diffusion-2-1-base"
new_pipe = StableDiffusionPipeline.from_pretrained(model_id).to("cuda")

# Function to generate image from text using the original model
def generate_image_from_text(text, pipe):
    with torch.cuda.amp.autocast():
        image = pipe(text).images[0]
    return image

# Function to generate image from text using the new model
def generate_image_from_text_new(text, pipe):
    generator = torch.Generator(device="cuda").manual_seed(42)
    with torch.cuda.amp.autocast():
        image = pipe(
            prompt=text,
            negative_prompt="Oversaturated, blurry, low quality",
            height=480,
            width=640,
            guidance_scale=8,
            num_inference_steps=35,
            generator=generator
        ).images[0]
    return image

# Function to generate video frames from text
def generate_frames(prompt, num_steps, batch_size, pipe):
    frames = []
    for _ in range(num_steps // batch_size):
        batch_frames = pipe(prompt, num_inference_steps=batch_size).frames
        frames.extend([frame for batch in batch_frames for frame in batch])
    return frames

# Function to export frames to a video file
def export_to_video(frames, output_path='output_video.mp4', fps=45):
    frames = [frame if isinstance(frame, np.ndarray) else np.array(frame) for frame in frames]
    frames = [(frame * 255).astype(np.uint8) if frame.dtype != np.uint8 else frame for frame in frames]
    height, width, layers = frames[0].shape
    size = (width, height)
    out = cv2.VideoWriter(output_path, cv2.VideoWriter_fourcc(*'mp4v'), fps, size)
    for frame in frames:
        out.write(cv2.cvtColor(frame, cv2.COLOR_RGB2BGR))
    out.release()
    return output_path

# Route to generate image using the original model
@app.route('/generate_image', methods=['POST'])
def generate_image():
    text_input = request.form['text']
    if not text_input:
        return jsonify(error="No prompt provided"), 400
    try:
        generated_image = generate_image_from_text(text_input, image_pipe)
        img_byte_arr = io.BytesIO()
        generated_image.save(img_byte_arr, format='PNG')
        img_byte_arr.seek(0)
        return send_file(img_byte_arr, mimetype='image/png', as_attachment=True, download_name='generated_image.png')
    except Exception as e:
        app.logger.error(f"Error generating image: {e}")
        return jsonify(error=str(e)), 500

# Route to generate image using the new model
@app.route('/generate_image_new', methods=['POST'])
def generate_image_new():
    text_input = request.form['text']
    if not text_input:
        return jsonify(error="No prompt provided"), 400
    try:
        generated_image = generate_image_from_text_new(text_input, new_pipe)
        img_byte_arr = io.BytesIO()
        generated_image.save(img_byte_arr, format='PNG')
        img_byte_arr.seek(0)
        return send_file(img_byte_arr, mimetype='image/png', as_attachment=True, download_name='generated_image.png')
    except Exception as e:
        app.logger.error(f"Error generating image: {e}")
        return jsonify(error=str(e)), 500

# Route to generate video
@app.route('/generate_video', methods=['POST'])
def generate_video():
    prompt = request.form.get('text')
    if not prompt:
        return jsonify(error="No prompt provided"), 400

    try:
        desired_video_length_seconds = 10
        fps = 10
        frames_needed = desired_video_length_seconds * fps
        batch_size = 45

        video_frames = generate_frames(prompt, frames_needed, batch_size, video_pipe)
        video_path = export_to_video(video_frames, fps=fps)

        with open(video_path, 'rb') as f:
            video_bytes = f.read()

        return send_file(io.BytesIO(video_bytes), mimetype='video/mp4', as_attachment=True, download_name='generated_video.mp4')
    except Exception as e:
        app.logger.error(f"Error generating video: {e}")
        return jsonify(error=str(e)), 500

# Function to start ngrok
def start_ngrok():
    public_url = ngrok.connect(9090)  # Use port 9090
    print(f" * ngrok tunnel \"{public_url}\" -> \"http://127.0.0.1:9090\"")
    return public_url

# Function to run Flask server
def run_flask():
    app.run(port=9090)

# Start ngrok and Flask server
ngrok.set_auth_token("2jND80KstRI4IZq0t0tKasn28LK_44K1Set5Lh4xvSPck7nHE")  # Replace with your ngrok authtoken
public_url = start_ngrok()
threading.Thread(target=run_flask).start()

# Print the public URL
print(f"Public URL: {public_url}")
```
